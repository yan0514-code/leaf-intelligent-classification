# 划分数据集

```python
import os
import shutil
import random

# 配置路径
root_path = r'C:\Users\27890\Desktop\叶子分类数据集'
classes = ["0", "1", "2"]
# 每类分配数量
split_num = {
    "0": {"train": 141, "val": 40, "test": 21},
    "1": {"train": 160, "val": 46, "test": 23},
    "2": {"train": 150, "val": 42, "test": 23}
}

# 创建目标文件夹
for split in ["train", "val", "test"]:
    for cls in classes:
        target_dir = os.path.join(root_path, split, cls)
        os.makedirs(target_dir, exist_ok=True)

# 分层复制图片
for cls in classes:
    cls_dir = os.path.join(root_path, cls)
    # 获取该类别所有jpg图片
    img_list = [f for f in os.listdir(cls_dir) if f.endswith(".jpg")]
    random.shuffle(img_list)  # 随机打乱顺序
    train_num = split_num[cls]["train"]
    val_num = split_num[cls]["val"]
    
    # 切分三段
    train_imgs = img_list[:train_num]
    val_imgs = img_list[train_num : train_num+val_num]
    test_imgs = img_list[train_num+val_num : ]

    # 复制到训练集
    for img in train_imgs:
        src = os.path.join(cls_dir, img)
        dst = os.path.join(root_path, "train", cls, img)
        shutil.copy(src, dst)
    # 复制到验证集
    for img in val_imgs:
        src = os.path.join(cls_dir, img)
        dst = os.path.join(root_path, "val", cls, img)
        shutil.copy(src, dst)
    # 复制到测试集
    for img in test_imgs:
        src = os.path.join(cls_dir, img)
        dst = os.path.join(root_path, "test", cls, img)
        shutil.copy(src, dst)

print("物理分层划分完成！")
```

# 数据预处理与增强

```python
import os
import random
import numpy as np
from PIL import Image, ImageFilter
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms

# ====================== 1. 自定义高斯轻度去噪变换 ======================
class LightGaussianBlur(object):
    """自定义轻度高斯模糊，仅去除反光噪点，保留叶片纹理边缘"""
    def __init__(self, radius=0.8):
        self.radius = radius  # 小半径，轻度平滑，不模糊叶片轮廓

    def __call__(self, img):
        # PIL高斯模糊
        return img.filter(ImageFilter.GaussianBlur(radius=self.radius))

# ====================== 2. 构建两套预处理/增强流水线 ======================
# 训练集transform：预处理全部操作 + 所有数据增强
train_transform = transforms.Compose([
    # 1.基础预处理
    transforms.Resize((512, 512)),       # 尺寸统一512*512
    LightGaussianBlur(radius=0.8),       # 轻度高斯去噪
    # 2.数据增强（仅训练集）
    transforms.RandomResizedCrop(size=512, scale=(0.75, 1.0)),  # 随机裁剪缩放
    transforms.RandomHorizontalFlip(p=0.5),                     # 随机水平翻转
    transforms.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.2), # 色彩抖动
    # 3.像素归一化两步
    transforms.ToTensor(),  # uint8(0-255) → float32(0~1)
    transforms.Normalize(mean=[0.485, 0.456, 0.406],
                         std=[0.229, 0.224, 0.225])
])

# 验证/测试集transform：仅基础预处理，无任何增强
val_test_transform = transforms.Compose([
    transforms.Resize((512, 512)),
    LightGaussianBlur(radius=0.8),  # 测试集同样做基础去噪
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406],
                         std=[0.229, 0.224, 0.225])
])

# ====================== 3. 自定义LeafDataset数据集类 ======================
class LeafDataset(Dataset):
    def __init__(self, root_dir, transform=None):
        self.root_dir = root_dir
        self.transform = transform
        self.image_paths = []
        self.labels = []

        # 扫描文件夹0、1、2，自动绑定标签
        class_names = sorted(os.listdir(root_dir))
        for label, cls_name in enumerate(class_names):
            cls_path = os.path.join(root_dir, cls_name)
            # 筛选jpg图片
            for img_name in os.listdir(cls_path):
                if img_name.lower().endswith(".jpg"):
                    full_path = os.path.join(cls_path, img_name)
                    self.image_paths.append(full_path)
                    self.labels.append(label)

    def __len__(self):
        # 重写__len__，返回样本总数
        return len(self.image_paths)

    def __getitem__(self, idx):
        # 重写__getitem__，按索引读取图像+标签
        img_path = self.image_paths[idx]
        label = self.labels[idx]
        # 通道转换：统一RGB三通道
        image = Image.open(img_path).convert("RGB")
        # 执行预处理/增强流水线
        if self.transform is not None:
            image = self.transform(image)
        return image, label

# ====================== 4. Mixup / CutMix 混合增强函数（训练循环调用） ======================
def mixup_data(x, y, alpha=0.2):
    """Mixup 图像混合增强"""
    if alpha > 0:
        lam = np.random.beta(alpha, alpha)
    else:
        lam = 1
    batch_size = x.size()[0]
    index = torch.randperm(batch_size)
    mixed_x = lam * x + (1 - lam) * x[index, :]
    y_a, y_b = y, y[index]
    return mixed_x, y_a, y_b, lam

def cutmix_data(x, y, alpha=0.2):
    """CutMix 区域混合增强"""
    if alpha > 0:
        lam = np.random.beta(alpha, alpha)
    else:
        lam = 1
    batch_size = x.size()[0]
    index = torch.randperm(batch_size)
    h, w = x.size()[2], x.size()[3]
    cut_rat = np.sqrt(1. - lam)
    cut_w = int(w * cut_rat)
    cut_h = int(h * cut_rat)
    cx = np.random.randint(w)
    cy = np.random.randint(h)
    x1 = max(cx - cut_w // 2, 0)
    y1 = max(cy - cut_h // 2, 0)
    x2 = min(cx + cut_w // 2, w)
    y2 = min(cy + cut_h // 2, h)
    mixed_x = x.clone()
    mixed_x[:, :, y1:y2, x1:x2] = x[index, :, y1:y2, x1:x2]
    y_a, y_b = y, y[index]
    lam = 1 - ((x2 - x1) * (y2 - y1) / (w * h))
    return mixed_x, y_a, y_b, lam

# Mixup损失计算
def mixup_criterion(criterion, pred, y_a, y_b, lam):
    return lam * criterion(pred, y_a) + (1 - lam) * criterion(pred, y_b)

# ====================== 5. 数据集实例化 + DataLoader ======================
if __name__ == '__main__':
    # 物理划分后的文件夹路径
    train_root = r'C:\Users\27890\Desktop\叶子分类数据集\train'
    val_root = r'C:\Users\27890\Desktop\叶子分类数据集\val'
    test_root = r'C:\Users\27890\Desktop\叶子分类数据集\test'

    # 构建数据集
    train_dataset = LeafDataset(train_root, transform=train_transform)
    val_dataset = LeafDataset(val_root, transform=val_test_transform)
    test_dataset = LeafDataset(test_root, transform=val_test_transform)

    # DataLoader加载器
    train_loader = DataLoader(
        train_dataset,
        batch_size=16,
        shuffle=True,
        num_workers=6,
        pin_memory=True
    )
    val_loader = DataLoader(
        val_dataset,
        batch_size=16,
        shuffle=False,
        num_workers=6,
        pin_memory=True
    )
    test_loader = DataLoader(
        test_dataset,
        batch_size=16,
        shuffle=False,
        num_workers=6,
        pin_memory=True
    )
```
